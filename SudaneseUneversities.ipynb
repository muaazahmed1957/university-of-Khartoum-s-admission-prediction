```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
lm = LinearRegression()


df = pd.read_csv("SudaneseCertificateData.csv")
df.head(12)
```

```python
df.drop(columns = ["number of requested" , "number of accepted"] , inplace = True)
df.head(12)
```

```python
df.drop([0,1,2,10,11,12],inplace = True)
df.head(10)
```

```python
# Update the values for 'number of setted' and 'number of passed' at index 9 (Year 2022)
df.loc[9, 'number of setted'] = 487010.0
df.loc[9, 'number of passed'] = 327173.0

# Calculate the percentage for 'pass%'
df.loc[9, 'pass%'] = (df.loc[9, 'number of passed'] / df.loc[9, 'number of setted']) * 100

df.head(7)
```

```python
sns.regplot(x="hundred's average", y="EEE UofK", data=df)
```

```python
df["hundred's average"].corr(df["EEE UofK"])
```

```python

correlation_hundred = df["hundred's average"].corr(df["EEE UofK"])
correlation_pass = df["pass%"].corr(df["EEE UofK"])

```

```python
print(correlation_hundred)
print(correlation_pass)
```
